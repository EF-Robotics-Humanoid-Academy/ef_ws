# Task 4 — Solved reference: custom audio-and-light mode

This mode only speaks and changes the headlight. It does not move the robot.


```python
from sdk_wrapper import G1

g1 = G1(iface="eth0", domain_id=0)

def enter_custom_mode():
    g1.say("entering custom mode")
    g1.set_headlight("blue", intensity=60, duration_s=5)
    print("Custom mode: audio announced and blue headlight enabled.")

enter_custom_mode()
```
