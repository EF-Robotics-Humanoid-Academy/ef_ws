## Introduction
The simplest useful voice-assistant loop: text in (English or German), a reply from the OpenAI API, printed and spoken back through `g1.say()`.

**Safety:** this task never moves the robot -- only speech. Construct `g1` the usual way regardless.

**Note:** `OPENAI_API_KEY` is already set in your account's environment; `openai_api_util.get_client()` picks it up automatically -- you never type the key yourself.

**Using Codex/AI here too:** these tasks build a chatbot step by step. If a cell is not obvious, paste the relevant `openai_api_util`/`sdk_wrapper.G1` function signature into Codex and ask it to write the cell, then read what it produced before running it.


```python
import sys
sys.path.append("..")
from sdk_wrapper import G1
import openai_api_util

g1 = G1(iface="eth0", domain_id=0)
client = openai_api_util.get_client()
```

## Task - one chat turn
`openai_api_util.chat_reply(client, text, history=history)` sends `text` plus the running `history` list to the model and returns its reply as a string. Append both the user's line and the reply to `history` yourself, so the next call still has context.

The default system prompt already tells the model to always answer in German, however it is asked -- that is why `g1.say(reply, language="DE")` below is safe to hardcode as the default.


```python
history = []

def chat_turn(text):
    reply = openai_api_util.chat_reply(client, text, history=history)
    history.append({"role": "user", "content": text})
    history.append({"role": "assistant", "content": reply})
    print(reply)
    g1.say(reply, language="DE")
    return reply
```

```python
chat_turn(input("Du: "))
```
