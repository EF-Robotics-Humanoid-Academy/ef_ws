# Task 16 — BrainCo Revo 2 pickup and delivery

This is the BrainCo-hand copy of Task 16. It keeps the SLAM/vision/arm sequence but uses the BrainCo hand helpers in `sdk_wrapper.py`.


## Safety and setup

Use a Python 3.9–3.12 kernel and install `bc_stark_sdk` in that kernel's Python environment. Confirm `device_index=0` is the intended Revo 2 hand. Keep all fingers clear. The hand opens before the arm reaches, closes only after the object is confirmed, and opens again at delivery.


```python
import json
import time
from openai_api_util import detect_object
from sdk_wrapper import G1

g1 = G1(iface="eth0", domain_id=0)
pose = json.loads(open("right_arm_forward_pose.json").read())

if detect_object("soda can"):
    g1.open_brainco_hand(device_index=0, duration_ms=1500)
    g1.interpolate_to_pose(pose)
    time.sleep(2)
    g1.close_brainco_hand(device_index=0, duration_ms=1500)
else:
    print("Object not detected; not moving the arm or hand.")

# Delivery/navigation steps belong here after a verified grasp.
# g1.navigate_to_point("dropoff")
g1.open_brainco_hand(device_index=0, duration_ms=1500)
g1.release_arms()
```
