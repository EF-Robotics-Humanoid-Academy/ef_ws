# Task 11 — Solved: BrainCo Revo 2 open and close

This reference uses the optional BrainCo integration in `sdk_wrapper.py`. Use Python 3.9–3.12 and install `bc_stark_sdk` first; the wrapper auto-detects the first hand as `device_index=0`.


```python
import time
from sdk_wrapper import G1

g1 = G1(iface="eth0", domain_id=0)
print(g1.open_brainco_hand(device_index=0, duration_ms=1500))
time.sleep(2)
print(g1.close_brainco_hand(device_index=0, duration_ms=1500))
time.sleep(2)
print(g1.open_brainco_hand(device_index=0, duration_ms=1500))
```
