In this short task you will run Unitree's supplied audio example from a Jupyter terminal. You do not need to write Python code.


## Steps

1. In Jupyter, open **File → New → Terminal**.
2. Copy and run these two commands, one at a time:

```bash
cd /home/unitree/unitree_sdk2_python/example/g1/audio
python g1_audio_client_example.py eth0
```

3. Listen for the robot's audio example. If it finishes without an error, the task is complete.

If `python` is not found, use `python3` for the second command. Do not change the interface name: this robot uses `eth0`.
