# Task 4 — Ask Codex for a custom audio-and-light mode

Make a small, safe custom mode. It must only announce a message and set the headlight; it must not move the robot.


## Steps

1. Open a Jupyter terminal in `~/academy/day_1`.
2. Ask Codex to create `custom_mode.py` with this prompt:

> Using the local `sdk_wrapper.py`, create `custom_mode.py`. Import `G1`, construct exactly one `G1(iface="eth0", domain_id=0)`, and define `enter_custom_mode()`. That function must call `g1.say("entering custom mode")` and then `g1.set_headlight("blue", intensity=60, duration_s=5)`. Print a short status message. Do not call any locomotion, arm, mode-switching, or developer-mode method. Run the function only under `if __name__ == "__main__":`.

3. Read the generated file. Verify that it contains only audio/light actions and no `walk_mode`, `prepare_mode`, `damp_mode`, `toggle_dev_mode`, or arm command.
4. With the area clear, run `python custom_mode.py`. The robot should say **“entering custom mode”** and show a blue headlight for about five seconds.
