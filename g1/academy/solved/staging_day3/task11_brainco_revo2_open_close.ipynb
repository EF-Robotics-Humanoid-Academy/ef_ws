# Task 11 — BrainCo Revo 2: open and close

This is the BrainCo Revo 2 version of the Dex3 hand exercise. It uses `sdk_wrapper.G1.open_brainco_hand()` and `close_brainco_hand()` rather than Unitree Dex3 DDS topics.


## Before you run it

- Use a Python 3.9–3.12 kernel: BrainCo does not support Python 3.8. Install the BrainCo `bc_stark_sdk` in that same notebook environment and ensure the current user can access the hand device (for serial, the user needs the `dialout` group and a new login).
- Connect only the intended Revo 2 hand and keep all fingers clear.
- The wrapper auto-detects BrainCo devices. `device_index=0` means the first detected hand.
- The BrainCo normalized position range is 0–1000: 0 opens and 1000 closes all six fingers.


```python
import time
from sdk_wrapper import G1

g1 = G1(iface="eth0", domain_id=0)

# Open first. This is also the safe final state.
g1.open_brainco_hand(device_index=0, duration_ms=1500)
time.sleep(2)

# Close only after confirming fingers and objects are clear.
g1.close_brainco_hand(device_index=0, duration_ms=1500)
time.sleep(2)

# Always leave the hand open at the end.
g1.open_brainco_hand(device_index=0, duration_ms=1500)
```
